```python
date_dir = [item for item in os.listdir(root_dir) if os.path.isdir(os.path.join(root_dir, item))]
```

```python
date_dir
```

```python
for date in date_dir:
    ID_list = [item for item in os.listdir(os.path.join(root_dir, date)) if os.path.isdir(os.path.join(root_dir, date, item))]
    
    clean_list = set([item.split('_')[0] for item in ID_list])
    print(f'Date {date}: {len(ID_list)}/{len(clean_list)}')
    
    pass_list = []
    
    for ID in clean_list:
        tmp = [item for item in ID_list if item.split('_')[0] == ID]
        left_right = [item.split('_')[1] for item in tmp]
        if not Counter(left_right) == Counter(['OD', 'OS']):
            print(ID, Counter(left_right))
        else: 
            pass_list.append(ID)
    pass_list.sort()
    print(pass_list)
```
